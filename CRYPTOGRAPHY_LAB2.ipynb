```python
import string

def encrypt_substitution(plaintext, key):
    alphabet = string.ascii_lowercase
    key = key.lower()
    ciphertext = ""

    for char in plaintext:
        if char.isalpha():
            idx = alphabet.index(char.lower())
            sub_char = key[idx]

            if char.isupper():
                ciphertext += sub_char.upper()
            else:
                ciphertext += sub_char
        else:
            ciphertext += char

    return ciphertext


def decrypt_substitution(ciphertext, key):
    alphabet = string.ascii_lowercase
    key = key.lower()
    plaintext = ""

    for char in ciphertext:
        if char.isalpha():
            idx = key.index(char.lower())
            sub_char = alphabet[idx]

            if char.isupper():
                plaintext += sub_char.upper()
            else:
                plaintext += sub_char
        else:
            plaintext += char

    return plaintext


print("=== Substitution Cipher Tool ===")
print("Standard Alphabet: abcdefghijklmnopqrstuvwxyz")

secret_key = input("Enter your 26-letter secret key: ")
message = input("Enter the text to encrypt: ")

if len(secret_key) != 26 or not secret_key.isalpha():
    print("Error: The key must be exactly 26 alphabetic characters.")
else:
    encrypted = encrypt_substitution(message, secret_key)
    print("\n[+] Encrypted Text:", encrypted)

    decrypted = decrypt_substitution(encrypted, secret_key)
    print("[+] Decrypted Text:", decrypted)
```
